```python
# A retail store offers a 20% discount if a customer’s total order exceeds $100
```

```python
Prd1 = int(input("Product 1 Rate:"))
Prd2 = int(input("Product 2 Rate:"))
Prd3 = int(input("Product 3 Rate:"))
add = Prd1+Prd2+Prd3
disc = (add*-0.2)
print("Net Value:",add)
if add>=100:
    print("Discount 20% :",disc)
    print("Gross Value:",add+disc)
else:
    print("Discount: 0")
    print("Gross Value",add)

    
```

```python
# Leap Year or normal year - Referred in Google
```

```python
year = int(input("Enter a year: "))

# Standard leap year rules
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```
