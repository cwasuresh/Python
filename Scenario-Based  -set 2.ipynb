```python
#  Write logic to find the factorial of a given number using recursion.
```

```python
def factorial(Num):
    if Num == 0 or Num == 1:
        return 1
    else:        
        return Num * factorial(Num - 1)
        
Num = int(input("Enter a number: "))
result = factorial(Num)
print("Factorial Value:",result)
```

```python
# Write logic to check whether a given number is an Armstrong number.
```

```python
def is_armstrong(num1):
    num_str = str(num1)
    num_digits = len(num_str)
    total_sum = sum(int(digit) ** num_digits for digit in num_str)
    return total_sum == num1

Num2 = int(input("Enter a Number:"))
if is_armstrong(Num2):
    print(f"{Num2} is an Armstrong number!")
else:
    print(f"{Num2} is not an Armstrong number.")
```

```python
# A password manager needs to strengthen weak passwords by swapping the first and last characters of user-generated passwords.
# Write logic to perform this operation on a given string.
```

```python
def strengthen_password(password):
    # If the password is empty or only 1 character, swapping has no effect
    if len(password) <= 1:
        return password
        
    # Take the last character, the middle section, and the first character
    return password[-1] + password[1:-1] + password[0]

# Example Usage:
user_password = input("Enter your password: ")
new_password = strengthen_password(user_password)

print(f"Original:  {user_password}")
print(f"Scrambled: {new_password}")
```

```python
# Write logic to find the longest word in a sentence.
```

```python
import string

def find_longest_word(sentence):
    # Remove punctuation so characters like '.' or ',' don't affect length
    clean_sentence = sentence.translate(str.maketrans('', '', string.punctuation))
    
    # Split the sentence into a list of individual words
    words = clean_sentence.split()
    
    # Handle empty input cases safely
    if not words:
        return ""
    
    # Find the maximum word based on its length
    return max(words, key=len)

# Example usage:
article_sentence = "Artificial intelligence tools help summarize complex articles efficiently."
print(find_longest_word(article_sentence))  # Output: intelligence
```

```python
# Write logic to check whether two given strings are anagrams.
```

```python
def is_anagram_sort(str1, str2):
    # Standardize both strings: remove spaces and convert to lowercase
    clean_str1 = "".join(str1.split()).lower()
    clean_str2 = "".join(str2.split()).lower()
    
    # If lengths don't match after cleaning, they cannot be anagrams
    if len(clean_str1) != len(clean_str2):
        return False
        
    # Sort and compare
    return sorted(clean_str1) == sorted(clean_str2)

# Example usage:
print(is_anagram_sort("Listen", "Silent"))          # Output: True
print(is_anagram_sort("Hello", "World"))            # Output: False
print(is_anagram_sort("Tom Marvolo Riddle", "I am Lord Voldemort")) # Output: True
```
