```python
# print Odd Numbers in the list

Numbers=(20,10,16,19,25,1,276,188)
for num in Numbers:
    if num % 2==1:
        print(num,"is odd")
```

```python
# print Even Numbers in the list

Numbers=(20,10,16,19,25,1,276,188)
for num in Numbers:
    if num % 2==0:
        print(num,"is even")
```
